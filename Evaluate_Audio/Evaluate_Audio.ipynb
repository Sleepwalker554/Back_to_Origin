```python
import pandas as pd

df = pd.read_csv("audio_quality_evaluation.csv")

methods = ['raw', 'mossformer', 'frcrn_se', 'demucs', 'denoiser', 'resemble']
metrics = ['ovrl', 'sig', 'bak']

avg_data = {}
for method in methods:
    avg_data[method] = {
        metric: df[f'{method}_{metric}'].mean() for metric in metrics
    }

avg_df = pd.DataFrame(avg_data).T
avg_df.columns = ['OVRL (Overall)', 'SIG (Signal)', 'BAK (Background)']
avg_df.index.name = 'Method'

print("=== DNSMOS 各方法平均分 ===\n")
print(avg_df.round(4).to_string())
avg_df.round(4)
```
